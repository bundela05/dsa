linear search

```python
def linearSearch(arr, target):
    for index,element in enumerate(arr):
        if element == target:
            return index
    return None
```

```python
data = [1,4,2,6,7,4,6,7]
```

```python
linearSearch(data,-7)
```

binary search

```python
def binarySearchVersion1(arr, target):
    low = 0
    high = len(arr)-1
    while low <= high:
        mid = low + ((high - low)//2)
        if arr[mid] == target:
            return mid
        if arr[mid] < target:
            low = mid + 1
        else:
            high = mid - 1
    return -1
```

```python
def binarySearchV2(arr, target , low = 0 , high = None):
    if high is None:
        high = len(arr)-1
        
    while low <= high:
        mid = low + ((high - low)//2)
        if arr[mid] == target:
            return mid
        if arr[mid] < target:
            low = mid + 1
        else:
            high = mid - 1
    return -1
```

exponential search

```python
def exponentialSearch(arr , target):
    if arr[0] == target:
        return 0

    i = 1
    n = len(arr)
    while (i<n) and (arr[i] <= target):
        i *= 2

    low = i // 2
    high = min(i, n-1)
    a = binarySearchV2(arr , target , low , high)
    return a
```

```python
exponentialSearch(data , 7)
```

```python
# find Largest no
no = [12,45,7,89,34]

# max(no)

lar = no[0]
for i in no[1:]:
    if i>lar:
        lar = i
print(f"largest element : {lar}")
```

```python
# count occ of ele
items = [1,2,4,4,3,5,5,3,6,4,5,3,3,2,5,6,6,5,3,2,3,2,4,5]
target = 2
count = 0
for i in items:
    if i == target:
        count += 1
    
print(count)
```

```python
# unique elements
new = []
for i in items:
    if i not in new :
        new.append(i)
print(new)
```

```python
# reverse 
# no[::-1]
no1 = no.copy() 
print(no1)
another = []

for i in range(len(no)):
    another.append(no1.pop())

print(another)
```

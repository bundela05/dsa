```python
class Node:
    def __init__(self , val):
        self.data = val
        self.next = None
        self.prev = None
```

```python

class CDLL:
    def __init__(self):
        self.head = None
        self.length = 0

    def isEmpty(self):
        return self.head is  None

    def display(self):
        if self.isEmpty():
            print("empty list")
            return
        current = self.head
        while True:
            print(current.data , end = " ")
            current = current.next
            if current == self.head:
                break
        print()

    def ins_beg(self, val):
        node = Node(val)
        if self.isEmpty():
            self.head = node
            node.next = node
            node.prev = node
            self.length += 1
            self.display()
            return

        current = self.head
        while current.next != self.head:
            current = current.next

        node.next = self.head
        node.prev = current
        self.head.prev = node
        current.next = node


        self.head = node

        self.length += 1
        self.display()

    def ins_end(self , val):
        node = Node(val)
        if self.isEmpty():
            self.head = node
            node.next = node
            node.prev = node
            self.length += 1
            self.display()
            return

        node.next = self.head
        node.prev = self.head.prev
        self.head.prev.next = node
        self.head.prev = node
        self.length += 1
        self.display()

```

```python
l1 = CDLL()
```

```python
l1.display()
```

```python
l1.ins_beg(13)
```

```python
l1.ins_end(13)
```

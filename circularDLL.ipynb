```python
class Node:
    def __init__(self , val):
        self.data = val
        self.next = None
        self.prev = None
```

```python

class CDLL:
    def __init__(self):
        self.head = None
        self.length = 0

    def isEmpty(self):
        return self.head is  None

    def display(self):
        if self.isEmpty():
            print("empty list")
            return
        current = self.head
        while True:
            print(current.data , end = " ")
            current = current.next
            if current == self.head:
                break
        print()

    def ins_beg(self, val):
        node = Node(val)
        if self.isEmpty():
            self.head = node
            node.next = node
            node.prev = node
            self.length += 1
            self.display()
            return

        current = self.head
        while current.next != self.head:
            current = current.next

        node.next = self.head
        node.prev = current
        self.head.prev = node
        current.next = node


        self.head = node

        self.length += 1
        self.display()

    def ins_end(self , val):
        node = Node(val)
        if self.isEmpty():
            self.head = node
            node.next = node
            node.prev = node
            self.length += 1
            self.display()
            return

        node.next = self.head
        node.prev = self.head.prev
        self.head.prev.next = node
        self.head.prev = node
        self.length += 1
        self.display()

    def ins_mid(self , pos , val):
        if pos < 0 or pos > self.length:
            print("check position argument")
            return

        if pos == 0:
            self.ins_beg(val)
            return
        if pos == self.length:
            self.ins_end(val)
            return

        node = Node(val)
        current = self.head
        for i in range(pos-1):
            current = current.next
        node.next = current.next
        node.prev = current
        current.next.prev = node
        current.next = node
        self.length += 1
        self.display()

    def del_beg(self):
        if self.isEmpty():
            print("nothing to delete")
            return
        if self.length == 1:
            print(f"{self.head.data} is deleted")
            self.head = None
            self.length -= 1
            self.display()
            return

        print(f"{self.head.data} is deleted")
        self.head.prev.next = self.head.next
        self.head.next.prev = self.head.prev
        self.head = self.head.next
        self.length -= 1
        self.display()

    def del_end(self):
        if self.isEmpty():
            print("nothing to delete")
            return
        if self.length == 1:
            print(f"{self.head.data} is deleted")
            self.head = None
            self.length -= 1
            self.display()
            return

        print(f"{self.head.prev.data} is deleted")
        self.head.prev = self.head.prev.prev
        self.head.prev.prev.next = self.head
        self.length -=1
        self.display()

    def del_by_index(self,pos):

        if pos <0 or pos> self.length - 1:
            print("invalid pos")
            return
        
        if self.isEmpty():
            print("nothing to delete")
            return
        if pos == 0:
            self.del_beg()
            return
        if pos == self.length-1:
            self.del_end()
            return
        current = self.head
        for i in range(pos):
            current = current.next
        print(f"{current.data} is deleted")
        current.prev.next = current.next
        current.next.prev = current.prev
        self.length -= 1
        self.display()



            

```

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.prev = None
        self.next = None


class CircularDoublyLinkedList:
    def __init__(self):
        self.head = None

    def prepend(self, data):
        """Insert at the beginning - O(1) time."""
        new_node = Node(data)
        if not self.head:
            new_node.next = new_node
            new_node.prev = new_node
            self.head = new_node
            return

        tail = self.head.prev

        # Wire new node between tail and old head
        new_node.next = self.head
        new_node.prev = tail

        # Update existing boundary nodes
        tail.next = new_node
        self.head.prev = new_node

        self.head = new_node

    def append(self, data):
        """Insert at the end - O(1) time."""
        # Appending is identical to prepending, except head doesn't shift
        self.prepend(data)
        self.head = self.head.next

    def delete(self, key):
        """Delete first occurrence of a value - O(n) time."""
        if not self.head:
            return

        curr = self.head

        while True:
            if curr.data == key:
                # Case 1: Only one node in the list
                if curr.next == curr:
                    self.head = None
                    return

                # Wire surrounding nodes together
                curr.prev.next = curr.next
                curr.next.prev = curr.prev

                # If deleting head, shift head to the next node
                if curr == self.head:
                    self.head = curr.next
                return

            curr = curr.next
            if curr == self.head:
                break  # Completed full circle, key not found

    def display_forward(self):
        """Traverse and print forward from head."""
        if not self.head:
            print("List is empty")
            return
```

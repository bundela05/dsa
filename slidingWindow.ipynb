sliding window technique is used on datastructure with continious memory like array,string

we can solve questions related to sub arrays init

1. maximum sum of subarray of size k

```python
def MaxSumSubArr(arr,k):
    result = float("-inf")
    window_sum = 0
    left = 0

    for right in range(len(arr)):
        window_sum += arr[right]

        if right - left +1 == k:
            result = max(result, window_sum)
            window_sum -= arr[left]
            left += 1
    
    return result
```

```python
MaxSumSubArr([2,1,3,5,4,6,0],3)
```

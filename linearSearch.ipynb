```python
def linear_search(arr: list, target: int) -> int:
    """
    Scans the array sequentially to find the target element.
    
    Returns:
        Index of the target if found, otherwise -1.
    """
    for index, value in enumerate(arr):
        if value == target:
            return index  # Target found at current index
    return -1  # Target not present in the array


# Example Usage
if __name__ == "__main__":
    numbers = [23, 45, 12, 8, 90, 56, 34]
    key = 90

    result = linear_search(numbers, key)

    if result != -1:
        print(f"Element {key} found at index: {result}")
    else:
        print(f"Element {key} not found in the array.")
```

recurseve approach

```python
def linear_search_recursive(arr: list, target: int, index: int = 0) -> int:
    """
    Recursively scans the array starting from the given index.
    """
    # Base Case 1: Reached the end without finding target
    if index >= len(arr):
        return -1
    
    # Base Case 2: Target found
    if arr[index] == target:
        return index
    
    # Recursive Case: Check the next element
    return linear_search_recursive(arr, target, index + 1)
```

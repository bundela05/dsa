```python
class Stack:
    def __init__(self):
        self.stack = []


    def isEmpty(self):
        return len(self.stack) == 0

    def display(self):
        if self.isEmpty():
            print("empty stack")
            return
        for i in self.stack[::-1]:
            print(i)
        print()

    def push(self ,val):
        self.stack.append(val)
        self.display()

    def pop(self):
        if self.isEmpty():
            print("empty stack")
            return None
        popped = self.stack.pop()
        print(popped)
    
    def peek(self):
        if self.isEmpty():
            print("empty stack")
            return None
        print(self.stack[-1])
```

```python
s1 = Stack()
```

```python
s1.push(10)
```

```python
s1.push(20)
```

```python
s1.display()
```

```python
s1.pop()
```

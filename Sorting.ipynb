sorting

```python
def swap(a,b):
    a = a+b
    b = a-b
    a = a-b
    return a,b
```

bubble sort

```python
def bubbleSort(arr):
    n = len(arr)
    for i in range(n):
        swapped = False
        for j in range(0,n-1-i):
            if arr[j] > arr[j+1]:
                arr[j],arr[j+1] = swap(arr[j],arr[j+1])
                # arr[j],arr[j+1] = arr[j],arr[j+1] works in python not in java ,cpp
                swapped = True
        if not swapped:
            break
    return arr
```

```python
arr = [3,5,1,2,4]
```

```python
bubbleSort(arr)
```

selection sort

```python
def selectionSort(arr):
    n = len(arr)
    for i in range(n-1):
        min_idx = i
        for j in range(i+1,n):
            if arr[min_idx] > arr[j]:
                min_idx = j
        arr[i],arr[min_idx] = arr[min_idx],arr[i]

    return arr
```

```python
arr2 = [1,2,3,9,6,3,1,2,8]
```

```python
selectionSort(arr2)
```

sorting

```python
def swap(a,b):
    a = a+b
    b = a-b
    a = a-b
    return a,b
```

bubble sort

```python
def bubbleSort(arr):
    n = len(arr)
    for i in range(n):
        swapped = False
        for j in range(0,n-1-i):
            if arr[j] > arr[j+1]:
                arr[j],arr[j+1] = swap(arr[j],arr[j+1])
                # arr[j],arr[j+1] = arr[j],arr[j+1] works in python not in java ,cpp
                swapped = True
        if not swapped:
            break
    return arr
```

```python
arr = [3,5,1,2,4]
```

```python
bubbleSort(arr)
```

selection sort

```python
def selectionSort(arr):
    n = len(arr)
    for i in range(n-1):
        min_idx = i
        for j in range(i+1,n):
            if arr[min_idx] > arr[j]:
                min_idx = j
        arr[i],arr[min_idx] = arr[min_idx],arr[i]

    return arr
```

```python
arr2 = [1,2,3,9,6,3,1,2,8]
```

```python
selectionSort(arr2)
```

insertion sort

```python
def insertionSort(arr):
    n = len(arr)
    for i in range(i,n):
        key = arr[i]
        j = i-1
        while(j >= 0)and(arr[j]>key):
            arr[j+1]= arr[j]
            j -= 1
        arr[j+1] = key
    return arr

```

merge sort

```python
def merge(left,right):
    result = []
    i = j = 0

    while((i<len(left)) and (j <len(right))):
        if left[i] < right[j]:
            result.append(left[i])
            i+=1
        else:
            result.append(right[j])
            j+=1

    result.extend(left[i:]) # for an uneven divide left has 3 and right has 2 then after comparison 1 elemnt is remaining in the list then .extend function adds that element to the and of list and other .extend function in our case list with 2 elements uses .extend([]) which does nothing
    result.extend(right[j:])
    
    return result

```

```python
def mergeSort(arr):
    if len(arr) == 1:
        return arr

    mid = len(arr)//2
    left = mergeSort(arr[:mid])
    right = mergeSort(arr[mid:])
    return merge(left,right)
```

quick sort

```python
def quickSort(arr):
    if len(arr) <= 1:
        return arr
    pivot = arr[len(arr)//2]

    left = [x for x in arr if x < pivot]
    middle = [x for x in arr if x == pivot]
    right = [x for x in arr if x > pivot]

    return quickSort(left) + middle + quickSort(right)
```

```python
# Example usage:
data = [38, 27, 43, 3, 9, 82, 10]
print(quickSort(data))  # Output: [3, 9, 10, 27, 38, 43, 82]
```

counting sort

```python
def counting_sort(arr):
    if not arr:
        return arr

    # 1. Find range to handle negative numbers
    min_val, max_val = min(arr), max(arr)
    range_span = max_val - min_val + 1

    # 2. Store counts of each element
    count = [0] * range_span
    for num in arr:
        count[num - min_val] += 1

    # 3. Transform count to store prefix sums (actual positions)
    for i in range(1, len(count)):
        count[i] += count[i - 1]

    # 4. Build output array stably (right to left)
    output = [0] * len(arr)
    for num in reversed(arr):
        idx = num - min_val
        count[idx] -= 1
        output[count[idx]] = num

    return output
```

```python

# Example usage:
data = [4, -2, 2, 8, 3, 3, 1]
print(counting_sort(data))  # Output: [-2, 1, 2, 3, 3, 4, 8]
```

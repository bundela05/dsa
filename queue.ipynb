```python
class QueueUsingList:
    def __init__(self):
        self.queue = []
    
    def isEmpty(self):
        return len(self.queue) == 0
    
    def display(self):
        if self.isEmpty():
            print("cannot display an empty queue")
            return
        print(self.queue)
    
    def enqueue(self, data):
        self.queue.append(data)
        self.display()

    def dequeue(self):
        if self.isEmpty():
            print("empty queue")
            return
        dequeuedElement = self.queue.pop(0)
        return dequeuedElement

    def frontpeek(self):
        if self.isEmpty():
            print("empty queue")
            return
        return self.queue[0]

    def rearpeek(self):
        if self.isEmpty():
            print("empty queue")
            return
        return self.queue[-1]
```

```python
q1 = QueueUsingList()
```

```python
q1.display()
```

```python
q1.enqueue(1)
q1.enqueue(2)
q1.enqueue(3)
q1.enqueue(4)
```

```python
q1.frontpeek()
```

```python
q1.rearpeek()
```

```python
q1.dequeue()
```

```python
q1.display()
```

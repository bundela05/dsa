queue using list

```python
class QueueUsingList:
    def __init__(self):
        self.queue = []
    
    def isEmpty(self):
        return len(self.queue) == 0
    
    def display(self):
        if self.isEmpty():
            print("cannot display an empty queue")
            return
        print(self.queue)
    
    def enqueue(self, data):
        self.queue.append(data)
        self.display()

    def dequeue(self):
        if self.isEmpty():
            print("empty queue")
            return
        dequeuedElement = self.queue.pop(0)
        return dequeuedElement

    def frontpeek(self):
        if self.isEmpty():
            print("empty queue")
            return
        return self.queue[0]

    def rearpeek(self):
        if self.isEmpty():
            print("empty queue")
            return
        return self.queue[-1]
```

```python
q1 = QueueUsingList()
```

```python
q1.display()
```

```python
q1.enqueue(1)
q1.enqueue(2)
q1.enqueue(3)
q1.enqueue(4)
```

```python
q1.frontpeek()
```

```python
q1.rearpeek()
```

```python
q1.dequeue()
```

```python
q1.display()
```

queue using linked list

```python
class Node:
    def __init__(self,data):
        self.data = data
        self.next = None
```

```python
class QueueUsingLL:
    def __init__(self):
        self.rear = None
        self.front = None

    def isEmpty(self):
        return self.front is None

    def display(self):
        if self.isEmpty():
            print("empty Queue")
            return
        curr = self.front
        while curr:
            print(curr.data, end= " ")
            curr = curr.next
        print()

    def enqueue(self,data):
        node = Node(data)
        if self.isEmpty():
            self.rear = self.front = node
        else:
            self.rear.next = node
            self.rear = node
        self.display()

    def dequeue(self):
        if self.isEmpty():
            print("empty queue")
            return
        if self.front == self.rear:
            print(f"dequed {self.front.data}")
            self.front = self.rear = None
            self.display()
            return
        print(f"dequed {self.front.data}")
        self.front = self.front.next
        self.display()

    def frontpeek(self):
        if self.isEmpty():
            print("empty queue")
            return
        return self.front.data
    
    def rearpeek(self):
        if self.isEmpty():
            print("empty queue")
            return
        return self.rear.data
```

using collections module

```python
from collections import deque
```

```python
class QueueUsingCollections:
    def __init__(self):
        self.queue = deque()

    def isEmpty(self):
        return len(self.queue) == 0

    def display(self):
        if self.isEmpty():
            print("empty queue")
        else:
            print(self.queue)

    def enqueue(self, data):
        self.queue.append(data)
        self.display()

    def dequeue(self):
        if self.isEmpty():
            print("empty queue")
            return
        dequeued = self.queue.popleft()
        self.display()
        return dequeued

    def frontPeek(self):
        if self.isEmpty():
            print("empty queue")
            return
        return self.queue[0]        

    def rearPeek(self):
        if self.isEmpty():
            print("empty queue")
            return
        return self.queue[-1]        
```

```python
q3.enqueue(1)
q3.enqueue(2)
q3.enqueue(13)
q3.enqueue(14)
```

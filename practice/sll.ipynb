```python
class Node:
    def __init__(self , data):
        self.data = data 
        self.next = None
```

```python
class SLL:
    def __init__(self):
        self.head = None
        self.length = 0

    def isEmpty(self):
        return self.length == 0

    def display(self):
        if self.isEmpty():
            print("empty linkedlist")
        else:
            current = self.head
            while current is not None:
                print(current.data ,end = " ")
                current = current.next
            print()

    def insert_begn(self,val):
            node = Node(val)
            if self.isEmpty():
                self.head = node
            else:
                node.next = self.head
                self.head = node
            self.length += 1
            self.display()

    def insert_end(self,val):
        node = Node(val)
        if self.isEmpty():
            self.head = node
        else :
            current = self.head
            while current.next != None:
                current = current.next
            current.next = node
        self.length += 1
        self.display()

    def insert_mid(self , pos , val):
        if (pos<0) or (pos>self.length):
            print("invalid value of position")
        elif pos == 0:
            self.insert_begn(val)
        elif self.length == (pos):
            self.insert_end(val)
        else:
            node = Node(val)
            current = self.head
            for i in range(pos - 1):
                current = current.next
            node.next = current.next
            current.next = node
            self.length += 1
            self.display()
    
    def search_index(self,val):
        if self.isEmpty():
            print("nothing to find in empty list")
            return -1
        
        else:
            count = 0
            current = self.head
            while current is not None:
                if current.data == val :
                    return count
                
                current = current.next
                count += 1
        
        print(f"Value {val} not found in list")
        return -1

        
    def del_beg(self):
        if self.isEmpty():
            print("nothing in linked list to delete")
        else:
            current = self.head
            print(f"deleted {current.data}")
            self.head = current.next
            self.length -= 1
            self.display()

    def del_end(self):
        if self.isEmpty():
            print("nothing in linked list to delete")
        
        elif self.length == 1:
            print(f"deleted {self.head.data}")
            self.head = None
            self.length -= 1
            self.display()
        
        else:
            current = self.head
            while current.next.next is not None:
                current = current.next
                print(current.next)
            current.next = None
            self.length -= 1
            print(f"deleted {current.data}")
            self.display()
    
    def del_by_index(self, pos):
        # 1. Check for invalid position or empty list
        if self.isEmpty() or pos < 0 or pos >= self.length:
            print("Invalid position or list is empty")
            return

        # 2. Special case: deleting the first node (index 0)
        if pos == 0:
            print(f"deleted {self.head.data}")
            self.head = self.head.next
            self.length -= 1
            self.display()
            return

        # 3. Traverse to the node EXACTLY ONE POSITION BEFORE the target
        current = self.head
        for _ in range(pos - 1):
            current = current.next  # Just move forward, do not modify pointers here

        # 4. 'current.next' is the node to delete. Bypass it.
        node_to_delete = current.next
        current.next = current.next.next
        
        self.length -= 1
        print(f"deleted {node_to_delete.data}")
        self.display()
```

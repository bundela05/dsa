bubble sort implimentation

```python
def bubbleSort(arr):
    n = len(arr)
    for i in range(n):
        swapped = False
        for j in range(n-i-1):
            if arr[j] > arr[j+1]:
                arr[j],arr[j+1] = arr[j+1],arr[j]
                swapped = True
        if swapped is False:
            break
    return arr
```

selection sort

```python
def selectionSort(arr):
    n = len(arr)
    for i in range(n-1):
        min_idx = i
        for j in range(i+1,n):
            if arr[min_idx]>arr[j]:
                min_idx = j
                print(f"min index{j}")
        if (min_idx != i):
            print(f"swapping {arr[i] , arr[min_idx]}")
            arr[i],arr[min_idx] = arr[min_idx],arr[i]
    return arr

            
```

insertion sort

```python
def insertionSort(arr):
    n  = len(arr)
    for i in range(1,n):
        j = i-1
        key = arr[i]
        while (j >= 0)and (arr[j] > key ):
            arr[j+1] = arr[j]
            j -= 1
        arr[j+1] = key      
    return arr
```

queue using stack

```python
from collections import deque
```

```python
class QueueUSingStack:
    def __init__(self):
            self.s1 = deque()
            self.s2 = deque()

    def enqueue(self, data):
        self.s1.append(data)

    def dequeue(self):
        if not self.s2:
            while self.s1:
                self.s2.append(self.s1.pop())
        if not self.s2:
             print("cannot dequeue from empty queue")
             return
        return self.s2.pop()
            
```

reverse first k in a queue

```python
que = deque([1,2,3,4,5])
k = 3
```

```python
def rev_first_k(que , k):
    stack = deque()
    result = deque()

    for _ in range(k):
        stack.append(que.popleft())

    while stack:
        result.append(stack.pop())

    while que:
        result.append(que.popleft())
    
    return result
```

```python
rev_first_k(que,k)
```

ticket counter

```python
from collections import deque


def get_last_person(n, k):
    if n == 1:
        return n

    # populating the ticket counter
    q = deque()
    for x in range(1, n + 1):
        q.append(x)

    while True:
        # remove the first k
        for _ in range(k):
            q.popleft()
            if len(q) == 1:
                return q[0]

        # remove the last k
        for _ in range(k):
            q.pop()
            if len(q) == 1:
                return q[0]
```

```python
get_last_person(9,3)
```

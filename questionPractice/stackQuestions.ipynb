stacks string reverse

```python
from collections import deque
```

```python
def rev_str(string):
    stack = deque()
    result = ""

    for ch in string:
        stack.append(ch)

    while stack:
        result += stack.pop()

    return result
```

```python
rev_str("happy birthday")
```

balanced parenthesis

```python
def isBalancedParenthesis(expression)-> bool:
    stack = deque()
    pairs = {
        ')':"(",
        '}':"{",
        ']':"["
    }
    for ch in expression:
        if ch in "({[":
            stack.append(ch)
        elif ch in ")}]":
            pair = pairs[ch]
            if not stack or pair != stack.pop():
                return False

    return len(stack) == 0
```

```python
def undoSimulation(commands):
    stack = deque()

    for command in commands:
        if command == "undo":
            if stack:
                stack.pop()
            else:
                print("empty stack cannot undo")
                
        else:
            stack.append(command[-1])
    return "".join(stack)
```

```python
commands = [
    "type A",
    "type B",
    "undo",
    "type C",
    "undo",
    "undo",
    "undo"
]
res = undoSimulation(commands)
print(f"result : {res}")
```

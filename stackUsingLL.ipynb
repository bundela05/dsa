```python
class Node:
    def __init__(self , data):
        self.data = data
        self.next = None
```

```python
class stackUSingLL:
    def __init__(self):
        self.top = None

    def isEmpty(self):
        return self.top is None

    def display(self):
        if self.isEmpty():
            print("cannot display")
            return
        current = self.top
        while current:
            print(current.data)
            current = current.next


    def push(self, data):
        node = Node(data)
        node.next = self.top
        self.top = node
        print(f"pushed {self.top.data}")

        self.display()

    def pop(self):
        if self.isEmpty():
            print("empty stack cannot pop")
            return
        popped = self.top.data
        self.top = self.top.next
        print(f"popped :{popped}")
```

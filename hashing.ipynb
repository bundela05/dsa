hashing

simple hash

```python
def simpleHash(key):
    total = 0
    for ch in key:
        total += ord(ch)
    return total
```

```python
simpleHash("ball"), simpleHash("bat")
```

multiplicative hash

```python
def multiplicativeHash(key):
    h = 1
    for ch in key:
        h = (h * 31)+ ord(ch)
    return h
```

```python
multiplicativeHash("ball")
```

DJB2

```python
def djb2(s):
    h = 5381
    for ch in s:
        (h << 31) + ord (ch)
    return h
```

```python
djb2("ball")
```

chained hashing (collision handelng)

```python
class chainingHashTable:
    def __init__(self , size = 5):
        self.size = size
        self.table =  [[] for _ in range(self.size)]

    def hashFun(self, key):
        return sum(ord(ch) for ch in key) % self.size

    def display(self):
        for i,bucket in enumerate(self.table):
            print(f"{i}: {bucket}")

    def insert(self, key, value = None):
        index = self.hashFun(key)
        for i,(k,v) in enumerate(self.table[index]):
            if k == key:
                self.table[index][i]=(key, value)
        self.table[index].append((key,value))

    def search(self, key): 
        index = self.hashFun(key)
        for (k,v)in self.table[index]:
            if k == key:
                return v
        return None
```

```python
a = chainingHashTable()
```

```python
a.display()
```

quadratic probing

```python
class LinearProbing:
    def __init__(self, size = 5):
        self.size = size
        self.table = [None]*self.size

    def hashFun(self,key):
        return sum(ord(ch) for ch in key)% self.size

    def display(self):
        print(self.table)
        print()

    def insert(self, key):
        h = self.hashFun(key)
        for i in range(self.size):
            index = (h+i)%self.size
            if self.table[index] is None or self.table[index] == key:
                self.table[index] = key
                return True
        print(f"Hash table is full cannot insert {key}")
        return False

    def search(self, key):
        h = self.hashFun(key)
        for i in range(self.size):
            index = (h+i)%self.size
            if self.table[index] is None:
                return False
            if self.table[index] == key:
                return index
        return False 
```

```python
a = LinearProbing()
```

```python
a.insert("ball")
a.insert("bat")
a.insert("baseball")
a.insert("blue")
```

hashing

simple hash

```python
def simpleHash(key):
    total = 0
    for ch in key:
        total += ord(ch)
    return total
```

```python
simpleHash("ball"), simpleHash("bat")
```

multiplicative hash

```python
def multiplicativeHash(key):
    h = 1
    for ch in key:
        h = (h * 31)+ ord(ch)
    return h
```

```python
multiplicativeHash("ball")
```

DJB2

```python
def djb2(s):
    h = 5381
    for ch in s:
        (h << 31) + ord (ch)
    return h
```

```python
djb2("ball")
```

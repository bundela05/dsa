```python
def three_sum(nums: list[int]) -> list[list[int]]:
    nums.sort()
    triplets = []
    n = len(nums)

    for i in range(n - 2):
        # Skip duplicate values for the first element
        if i > 0 and nums[i] == nums[i - 1]:
            continue

        left, right = i + 1, n - 1
        while left < right:
            total = nums[i] + nums[left] + nums[right]

            if total < 0:
                left += 1
            elif total > 0:
                right -= 1
            else:
                triplets.append([nums[i], nums[left], nums[right]])

                # Skip duplicates for the second and third elements
                while left < right and nums[left] == nums[left + 1]:
                    left += 1
                while left < right and nums[right] == nums[right - 1]:
                    right -= 1

                left += 1
                right -= 1

    return triplets


# Example usage:
nums = [-1, 0, 1, 2, -1, -4]
print(three_sum(nums))
# Output: [[-1, -1, 2], [-1, 0, 1]]
```

```python
def two_sum(arr, target):
    left, right = 0, len(arr) - 1
    while left < right:
        s = arr[left] + arr[right]
        if s == target:
            return (left, right)
        elif s < target:
            left += 1
        else:
            right -= 1
    return []

two_sum([2, 7, 11, 15], 9)
```

```python
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

def has_cycle(head: ListNode) -> bool:
    slow = fast = head

    while fast and fast.next:
        slow = slow.next          # Move 1 step
        fast = fast.next.next     # Move 2 steps

        if slow == fast:          # Cycle detected
            return True

    return False                  # Reached null end
```

2. find the start of the cycle in the linked list

```python
class Solution:
    def __init__(self):
        self.slow = None
        self.fast = None

    def has_cycle(self, head):
        self.slow, self.fast = head, head

        while self.fast and self.fast.next:
            self.slow = self.slow.next
            self.fast = self.fast.next.next
            if self.slow == self.fast:
                return True
        return False

    def start_cycle(self,head):
        if not self.has_cycle(head):
             return None

        self.slow = head
        while self.slow != self.fast:
            self.slow = self.slow.next
            self.fast = self.fast.next

        return self.slow
```

3. find duplicate Number (floyed cycle detection on arrays)

```python
def DuplicateDetection(arr):
    slow = arr[0]
    fast = arr[arr[0]]

    while slow != fast:
        slow = arr[slow]
        fast = arr[arr[fast]]

    slow = 0
    while slow != fast:
        slow = arr[slow]
        fast = arr[fast]

    return slow
```

```python
def subArrSum(arr , k):
    count = 0
    prefixSum = 0
    freq = {0 : 1}

    for num in arr:
        prefixSum += num
        if prefixSum - k in freq:
            count += freq[prefixSum - k]

        freq[prefixSum] = freq.get(prefixSum, 0 )+1

    return count
```

```python
def longestSubArr(arr):
    prefixSum = 0 
    longest = 0
    first_occ = {0: -1}

    for i,num in enumerate(arr):
        prefixSum += 1 if num == 1 else -1

        if prefixSum in first_occ:
            longest = max(longest,i -first_occ[prefixSum])
        else:
            first_occ[prefixSum] = i

    return longest
```

```python
def subarraysDivByK(nums: list[int], k: int) -> int:
    # Dictionary to store frequency of prefix sum remainders
    # Initialize with 0: 1 to account for subarrays starting from index 0
    remainder_counts = {0: 1}
    
    prefix_sum = 0
    count = 0
    
    for num in nums:
        prefix_sum += num
        
        # Calculate positive remainder
        remainder = prefix_sum % k
        
        # If this remainder has been seen before, add its frequency to count
        if remainder in remainder_counts:
            count += remainder_counts[remainder]
            remainder_counts[remainder] += 1
        else:
            remainder_counts[remainder] = 1
            
    return count

# Example Usage
nums = [4, 5, 0, -2, -3, 1]
k = 5
print(subarraysDivByK(nums, k))  # Output: 7
```

```python
from collections import Counter


def topKFrequent(nums: list[int], k: int) -> list[int]:
  # Step 1: Count the frequency of each number
  count = Counter(nums)

  # Step 2: Create buckets where index = frequency
  # Max possible frequency is len(nums)
  buckets = [[] for _ in range(len(nums) + 1)]
  for num, freq in count.items():
    buckets[freq].append(num)

  # Step 3: Iterate through buckets in reverse to get top k elements
  result = []
  for freq in range(len(buckets) - 1, 0, -1):
    for num in buckets[freq]:
      result.append(num)
      if len(result) == k:
        return result


# Example Usage:
nums = [1, 1, 1, 2, 2, 3]
k = 2
print(topKFrequent(nums, k))  # Output: [1, 2]
```

```python
def is_anagram(s: str, t: str) -> bool:
    if len(s) != len(t):
        return False

    count_s = {}
    count_t = {}

    for i in range(len(s)):
        count_s[s[i]] = count_s.get(s[i], 0) + 1
        count_t[t[i]] = count_t.get(t[i], 0) + 1

    return count_s == count_t


# Example from the screenshot
print(is_anagram("hand", "band"))  # Output: False
print(is_anagram("anagram", "nagaram"))  # Output: True
```

```python
def is_isomorphic(s: str, t: str) -> bool:
    if len(s) != len(t):
        return False

    map_s_to_t = {}
    map_t_to_s = {}

    for char_s, char_t in zip(s, t):
        # Check s -> t mapping
        if char_s in map_s_to_t and map_s_to_t[char_s] != char_t:
            return False
        # Check t -> s mapping
        if char_t in map_t_to_s and map_t_to_s[char_t] != char_s:
            return False

        map_s_to_t[char_s] = char_t
        map_t_to_s[char_t] = char_s

    return True


# Example usage
if __name__ == "__main__":
    print(is_isomorphic("egg", "add"))  # True
    print(is_isomorphic("foo", "bar"))  # False
    print(is_isomorphic("paper", "title"))  # True
```

```python
def search_range(nums: list[int], target: int) -> list[int]:
    def find_bound(is_first: bool) -> int:
        low, high = 0, len(nums) - 1
        bound = -1
        
        while low <= high:
            mid = low + (high - low) // 2
            
            if nums[mid] == target:
                bound = mid
                if is_first:
                    high = mid - 1  # Keep searching left
                else:
                    low = mid + 1   # Keep searching right
            elif nums[mid] < target:
                low = mid + 1
            else:
                high = mid - 1
                
        return bound

    first = find_bound(is_first=True)
    # Optimization: if target isn't found at all, skip the second search
    if first == -1:
        return [-1, -1]
        
    last = find_bound(is_first=False)
    return [first, last]

# Example usage
arr = [5, 7, 7, 8, 8, 10]
print(search_range(arr, 8))  # Output: [3, 4]
print(search_range(arr, 6))  # Output: [-1, -1]
```

```python
from typing import List


class Solution:

    def minDays(self, bloomDay: List[int], m: int, k: int) -> int:
        # Edge case: Not enough total flowers to make m bouquets of k flowers
        if m * k > len(bloomDay):
            return -1

        # Helper function to check if we can make 'm' bouquets by 'day'
        def canMakeBouquets(day: int) -> bool:
            bouquets = 0
            adjacent_flowers = 0

            for bloom in bloomDay:
                if bloom <= day:
                    adjacent_flowers += 1
                    if adjacent_flowers == k:
                        bouquets += 1
                        adjacent_flowers = (
                            0  # Reset counter after forming a bouquet
                        )
                else:
                    adjacent_flowers = (
                        0  # Reset counter if a flower hasn't bloomed
                    )

            return bouquets >= m

        # Binary search between the earliest and latest bloom day
        left = min(bloomDay)
        right = max(bloomDay)
        ans = -1

        while left <= right:
            mid = left + (right - left) // 2

            if canMakeBouquets(mid):
                ans = mid  # Valid day found, try to find an even smaller day
                right = mid - 1
            else:
                left = mid + 1  # Not enough bouquets, need to wait more days

        return ans


# --- Example Usage ---
if __name__ == "__main__":
    sol = Solution()
    bloomDay = [1, 10, 3, 10, 2]
    m = 3
    k = 1
    print(f"Minimum Days: {sol.minDays(bloomDay, m, k)}")  # Output: 3
```

```python
def can_place_cows(stalls, k, min_dist):
    """Helper function to check if k cows can be placed with at least min_dist apart."""
    # Place the first cow in the first stall
    cows_placed = 1
    last_position = stalls[0]
    
    for i in range(1, len(stalls)):
        # If the current stall is at least min_dist away from the last placed cow
        if stalls[i] - last_position >= min_dist:
            cows_placed += 1
            last_position = stalls[i]
            
            # If all cows are successfully placed, return True
            if cows_placed == k:
                return True
                
    return False

def aggressive_cows(stalls, k):
    # Step 1: Sort the stall positions to apply binary search
    stalls.sort()
    
    # Step 2: Define the search space for the minimum distance
    low = 1
    high = stalls[-1] - stalls[0]
    ans = 0
    
    # Step 3: Binary search on the answer
    while low <= high:
        mid = (low + high) // 2
        
        if can_place_cows(stalls, k, mid):
            ans = mid        # Valid distance found, try for a larger one
            low = mid + 1
        else:
            high = mid - 1   # Distance too large, try a smaller one
            
    return ans

# --- Example Usage ---
if __name__ == "__main__":
    stall_positions = [1, 2, 8, 4, 9]
    num_cows = 3
    
    result = aggressive_cows(stall_positions, num_cows)
    print(f"Largest minimum distance: {result}")  # Output: 3 (Cows at 1, 4, 8 or 1, 4, 9)
```

```python
def activity_selection(start: list[int], end: list[int]) -> int:
    # 1. Pair start and end times, then sort by end time (x[1])
    activities = sorted(zip(start, end), key=lambda x: x[1])
    
    if not activities:
        return 0
    
    # 2. Always select the first activity (earliest finish time)
    count = 1
    last_end = activities[0][1]
    
    # 3. Greedily select subsequent compatible activities
    for current_start, current_end in activities[1:]:
        # Use >= if one activity can start at the exact moment the previous ends
        if current_start >= last_end:
            count += 1
            last_end = current_end
            
    return count


# Example Test Case
start_times = [1, 3, 0, 5, 8, 5]
end_times   = [2, 4, 6, 7, 9, 9]

print("Maximum activities:", activity_selection(start_times, end_times))
# Output: 4 -> Selected: (1, 2), (3, 4), (5, 7), (8, 9)
```

```python
def find_min_platforms(arr, dep):
    """
    Finds the minimum number of railway platforms required.
    
    Parameters:
    arr (list): Arrival times of trains
    dep (list): Departure times of trains
    
    Returns:
    int: Minimum platforms needed
    """
    n = len(arr)
    
    # Sort arrival and departure arrays independently
    arr.sort()
    dep.sort()
    
    platforms_needed = 0
    max_platforms = 0
    
    i = 0  # Pointer for arrival array
    j = 0  # Pointer for departure array
    
    # Traverse both arrays
    while i < n and j < n:
        # If next event is an arrival (or simultaneous arrival & departure),
        # an extra platform is needed
        if arr[i] <= dep[j]:
            platforms_needed += 1
            max_platforms = max(max_platforms, platforms_needed)
            i += 1
        # Else a train departs, freeing up a platform
        else:
            platforms_needed -= 1
            j += 1
            
    return max_platforms

# Example Usage (times in 24-hour HHMM format)
arrivals = [900, 940, 950, 1100, 1500, 1800]
departures = [910, 1200, 1120, 1130, 1900, 2000]

print("Minimum Platforms Required:", find_min_platforms(arrivals, departures))
# Output: 3
```

```python
def canJump(nums):
    max_reach = 0
    for i, jump in enumerate(nums):
        if i > max_reach:
            return False
        max_reach = max(max_reach, i + jump)
        if max_reach >= len(nums) - 1:
            return True
    return True

print(canJump([2, 3, 1, 1, 4]))  # Output: True
print(canJump([3, 2, 1, 0, 4]))  # Output: False
```

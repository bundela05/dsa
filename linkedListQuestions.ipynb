```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None
        self.prev = None


def find_middle_cdll(head):
    """
    Finds the middle node of a Circular Doubly Linked List.
    - Time Complexity: O(N) (specifically N / 2 steps)
    - Space Complexity: O(1) auxiliary space
    """
    if head is None:
        return None

    # Single-node list
    if head.next == head:
        return head

    left = head
    right = head.prev  # CDLL tail is directly accessible via head.prev

    # Traverse towards each other until they meet or cross
    while left != right and left.next != right:
        left = left.next
        right = right.prev

    # For odd length: left == right
    # For even length: left.next == right -> returns first middle (or 'right' for second middle)
    return left
```

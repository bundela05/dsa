```python
# anagrams

s1 = "silent"
s2 = "listen"

sorted(s1) == sorted(s2)
```

```python
# way2

l = [0]*26

for ch in s1:
    l[ord(ch)-97] += 1

for ch in s2:
    l[ord(ch)-97] -= 1
if all(m==0 for m in l):
    print("anagram")

else :
    print("not anagram")
```

```python
class Node:
    def __init__(self,val):
        self.prev = None
        self.data = val
        self.next = None
```

```python
class DLL:
    def __init__(self):
        self.head = None
        self.length = 0

    def isEmpty(self):
        return self.head is None

    def display(self):
        if self.isEmpty():
            print("empty linked list")
        else:
            current = self.head
            while current is not None:
                print(current.data , end = " ")
                current = current.next

    def insert_beg(self , val):
        node = Node(val)
        if self.isEmpty():
            self.head = node
            self.length += 1
            return

        node.next = self.head
        self.head.prev = node
        self.head = node 
        self.length += 1
        self.display()

    def insert_end(self , val):
        node = Node(val)
        if self.isEmpty():
            self.head = node
            self.length += 1
            return
        
        current = self.head
        while current.next is not None:
            current = current.next
        node.prev = current
        current.next = node
        self.length += 1
        self.display()

    def insert_mid(self , pos , val):
        if pos< 0 or pos > self.length:
            print("invalid pos")
            return
        
        if pos == 0:
            self.insert_beg(val)
            return
        if pos == (self.length):
            self.insert_end(val)
            return
        
        node = Node(val)
        current = self.head
        for i in range(pos - 1):
            current = current.next

        node.next = current.next
        node.prev = current
        current.next.prev = node
        current.next = node
        self.length += 1
        self.display()

    def del_beg(self):
        if self.isEmpty():
            print("nothing in list to be deleted")
            return
        if self.length == 1:
            self.head = None
            self.length -= 1
            self.display()
            return
        self.head.next.prev = None
        self.head = self.head.next
        self.length -= 1
        self.display()

```

factorial

```python
def factorial(n):
    if n == 1 or n==0:
        return 1
    return n*factorial(n-1)
```

```python
factorial(5)
```

fibonacci

```python
def fibonacci(n):
    if n == 0 :
        return 0
    if n == 1:
        return 1
    return fibonacci(n-1)+fibonacci(n-2)
```

```python
fibonacci(5)
```

sum of first n natural numbers

```python
def sumOfN(n):
    if n <= 1:
        return 1
    return n + sumOfN(n-1)
```

```python
sumOfN(3)
```

count digits of a number

```python
def countDigits(n):
    if n == 0:
        return 0
    return 1 + countDigits(n//10)
```

```python
countDigits(1235)
```

 power function

```python
def getPower(n , k ):
    if k == 0:
        return 1
    return n* getPower(n,k-1)
```

```python
getPower(4,5)
```
